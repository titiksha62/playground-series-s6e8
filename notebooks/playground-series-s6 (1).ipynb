```python
import pandas as pd

train = pd.read_csv("/kaggle/input/competitions/playground-series-s6e8/train.csv")
test = pd.read_csv("/kaggle/input/competitions/playground-series-s6e8/test.csv")

print("Train shape:", train.shape)
print("Test shape:", test.shape)

train.head()
```

```python
# Check column names
print("Train columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())
```

```python
# Check missing values
print("Missing values in train:")
print(train.isnull().sum())

print("\nMissing values in test:")
print(test.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate rows in train:", train.duplicated().sum())
print("Duplicate rows in test:", test.duplicated().sum())
```

```python
# Check unique values in every column
for col in train.columns:
    print(f"\n{col}:")
    print("Unique values:", train[col].nunique())
    print(train[col].unique()[:10])
```

```python
# Compare numerical features between addicted and non-addicted users

numeric_cols = train.select_dtypes(include="number").columns.drop(
    ["id", "addicted_label"]
)

train.groupby("addicted_label")[numeric_cols].mean().T
```

```python
# Target distribution for categorical features

categorical_cols = ["gender", "stress_level", "academic_work_impact"]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(
        pd.crosstab(
            train[col],
            train["addicted_label"],
            normalize="index"
        ).round(3)
    )
```
